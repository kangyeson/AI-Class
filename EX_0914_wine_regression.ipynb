```python
import pandas as pd
from sklearn.model_selection import train_test_split

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/wine.csv")
```

```python
prevX = df.drop("Wine", axis=1)
X = prevX.drop("Alcohol", axis=1)
y = df["Alcohol"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

Linear

```python
model = LinearRegression()
model.fit(X_train, y_train)
ly_preds = model.predict(X_test)
```

```python
print('평균제곱오차', mean_squared_error(ly_preds, y_test))
```

```python
def mse_np(actual, predicted):
    return np.mean((np.array(actual) - np.array(predicted)) ** 2)

print('평균제곱오차', mse_np(ly_preds, y_test))
```

```python
def mse(actual, predicted):
    sum_square_error = sum((a - p) ** 2 for a, p in zip(actual, predicted))
    mean_square_error = sum_square_error / len(actual)
    return mean_square_error

print('평균제곱오차', mse(ly_preds, y_test))
```

DecisionTree

```python
from sklearn.tree import DecisionTreeRegressor

decision_model = DecisionTreeRegressor()
decision_model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error

dy_preds = decision_model.predict(X_test)
print('평균제곱오차', mean_squared_error(dy_preds, y_test))
```

RandomForest


```python
from sklearn.ensemble import RandomForestRegressor

random_model = RandomForestRegressor()
random_model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error

ry_preds = random_model.predict(X_test)
print('평균제곱오차', mean_squared_error(ry_preds, y_test))
```

SVR


```python
from sklearn.svm import SVR

svr_model = SVR(kernel='linear')
svr_model.fit(X_train, y_train)

from sklearn.metrics import mean_squared_error

svr_preds = svr_model.predict(X_test)
print('평균제곱오차', mean_squared_error(svr_preds, y_test))
```
